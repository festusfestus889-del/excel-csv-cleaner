```python
import pandas as pd
import numpy as np
from datetime import datetime
print("Libraries loaded")
```

```python
# Create messy data like real client data
data = {
    'Customer Name': [' JOHN DOE ', 'jane smith', None, ' JOHN DOE ', 'Chidi Okoro', 'AMAKA ', ''],
    'Email': ['JOHN@GMAIL.COM', 'jane@gmail.com ', 'chidi@YAHOO.COM', 'JOHN@GMAIL.COM', None, 'amaka@gmail.com', 'test@GMAIL.com'],
    'Order Date': ['12/05/2023', '2023.05.12', '05-13-2023', '12/05/2023', '2023/05/14', '14-05-2023', ''],
    'Amount': ['1000', '2500.50', None, '1000', ' 3000 ', 'invalid', '1500'],
    'Region': ['lagos', 'Lagos ', 'ABUJA', 'lagos', 'PH', 'abuja', 'Lagos']
}

df = pd.DataFrame(data)
df.to_csv('messy_data.csv', index=False)
print("Messy file created")
df.head(10)
```

```python
def clean_dataset(file_path):
    # Load file
    df = pd.read_csv(file_path)
    print(f"BEFORE: {df.shape[0]} rows")
    print(df.head())

    # 1. Clean Customer Name: strip spaces, title case, remove empty
    df['Customer Name'] = df['Customer Name'].astype(str).str.strip().str.title()
    df = df[~df['Customer Name'].isin(['', 'None', 'Nan'])]

    # 2. Clean Email: lowercase and strip
    df['Email'] = df['Email'].astype(str).str.strip().str.lower()
    df.loc[df['Email'].isin(['none', 'nan', '']), 'Email'] = np.nan

    # 3. Clean Order Date: handle multiple formats
    df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce', dayfirst=True)

    # 4. Clean Amount: convert to number, remove invalid
    df['Amount'] = pd.to_numeric(df['Amount'].astype(str).str.strip(), errors='coerce')

    # 5. Clean Region: standardize
    df['Region'] = df['Region'].astype(str).str.strip().str.title()
    df['Region'] = df['Region'].replace({'Ph': 'Port Harcourt', 'Abuja': 'Abuja', 'Lagos': 'Lagos'})

    # 6. Remove duplicates
    df = df.drop_duplicates()

    # 7. Drop rows where critical data is missing
    df = df.dropna(subset=['Customer Name', 'Amount'])

    print(f"\nAFTER: {df.shape[0]} rows")
    print(df.head())

    return df

clean_df = clean_dataset('messy_data.csv')
```

```python
# Export to Excel and CSV for client
clean_df.to_csv('clean_data.csv', index=False)
clean_df.to_excel('clean_data.xlsx', index=False)

print("Clean files created! Ready for GitHub.")
clean_df
```
